```python
# -*- coding: utf-8 -*-
"""
Minh họa tối ưu ngữ nghĩa (Semantic), sách Chương 19, mục 19.10 (trang 737 đến 738).
Dùng ràng buộc trên lược đồ để viết lại hoặc bỏ bớt truy vấn.
"""

# ----- DỮ LIỆU GIẢ LẬP -----
DEPARTMENT = [
    {"Dnumber": 1, "Dname": "Headquarters"},
    {"Dnumber": 4, "Dname": "Administration"},
    {"Dnumber": 5, "Dname": "Research"},
]
EMPLOYEE = [  # Super_ssn = mã người giám sát, Dno = mã phòng ban (None = chưa có phòng)
    {"Ssn": 1, "Lname": "Borg",    "Salary": 55000, "Super_ssn": None, "Dno": 1},
    {"Ssn": 2, "Lname": "Wong",    "Salary": 40000, "Super_ssn": 1,    "Dno": 5},
    {"Ssn": 3, "Lname": "Smith",   "Salary": 30000, "Super_ssn": 2,    "Dno": 5},
    {"Ssn": 4, "Lname": "Wallace", "Salary": 43000, "Super_ssn": 1,    "Dno": 4},
    {"Ssn": 5, "Lname": "Zelaya",  "Salary": 25000, "Super_ssn": 4,    "Dno": 4},
    {"Ssn": 6, "Lname": "Nguyen",  "Salary": 38000, "Super_ssn": 2,    "Dno": None},
    {"Ssn": 7, "Lname": "Tran",    "Salary": 26000, "Super_ssn": 4,    "Dno": 4},
]
theo_ssn = {e["Ssn"]: e for e in EMPLOYEE}


def in_bang(ten, rows):
    print(f"\n{ten}: {len(rows)} dòng")
    if rows:
        cols = list(rows[0])
        print("   " + " | ".join(f"{c:<10}" for c in cols))
        for r in rows:
            print("   " + " | ".join(f"{str(r[c]):<10}" for c in cols))


print("=== DỮ LIỆU ===")
in_bang("EMPLOYEE", EMPLOYEE)
in_bang("DEPARTMENT", DEPARTMENT)

# ===================== VÍ DỤ 1 =====================
print("\n\n=== VÍ DỤ 1: nhân viên lương cao hơn người giám sát ===")
print("Ràng buộc trên lược đồ: không nhân viên nào lương cao hơn người giám sát.")
print("\nKiểm tra từng nhân viên so với người giám sát:")
ket_qua = []
for e in EMPLOYEE:
    if e["Super_ssn"] is None:
        continue
    sep = theo_ssn[e["Super_ssn"]]
    cao_hon = e["Salary"] > sep["Salary"]
    print(f"   {e['Lname']:<8} {e['Salary']:>6}  so với  {sep['Lname']:<8} {sep['Salary']:>6}  -> cao hơn? {cao_hon}")
    if cao_hon:
        ket_qua.append((e["Lname"], sep["Lname"]))
print(f"\nKết quả chạy thật: {len(ket_qua)} dòng")
print("Bộ tối ưu ngữ nghĩa biết trước kết quả rỗng nhờ ràng buộc, nên KHÔNG cần chạy truy vấn.")

# ===================== VÍ DỤ 2 =====================
print("\n\n=== VÍ DỤ 2: họ và lương của nhân viên lương > 35000, có nối với DEPARTMENT ===")
print("Ràng buộc: EMPLOYEE.Dno là khóa ngoại tham chiếu DEPARTMENT.Dnumber.")

so_sanh = 0
goc = []
for e in EMPLOYEE:
    for d in DEPARTMENT:
        so_sanh += 1
        if e["Dno"] == d["Dnumber"] and e["Salary"] > 35000:
            goc.append({"Lname": e["Lname"], "Salary": e["Salary"]})
in_bang("Câu gốc (có nối DEPARTMENT)", goc)
print(f"Số lần so sánh: {len(EMPLOYEE)} nhân viên x {len(DEPARTMENT)} phòng ban = {so_sanh}")

moi = [{"Lname": e["Lname"], "Salary": e["Salary"]} for e in EMPLOYEE
       if e["Dno"] is not None and e["Salary"] > 35000]
in_bang("Câu đã biến đổi (bỏ DEPARTMENT, thêm Dno IS NOT NULL)", moi)
print(f"Số lần so sánh: {len(EMPLOYEE)} (chỉ duyệt EMPLOYEE)")

sai = [{"Lname": e["Lname"], "Salary": e["Salary"]} for e in EMPLOYEE if e["Salary"] > 35000]
in_bang("Thử bỏ điều kiện IS NOT NULL (để thấy vì sao cần nó)", sai)

print("\nHai câu đúng cho cùng kết quả:", goc == moi)
print("Thiếu IS NOT NULL thì sai: có thêm Nguyen (chưa có phòng ban) mà câu gốc đã loại.")
```
